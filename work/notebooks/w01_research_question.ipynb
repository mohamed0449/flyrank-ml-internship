### Quick Look at the Data

The dataset contains **30,000** content pages. About **54.2%** of the pages show a downward performance trend, suggesting that many pages may require attention. In addition, **21.2%** of the pages are more than one year old, indicating that a substantial portion of the content may benefit from review or updating. These observations support choosing the **Content Refresh / Content Opportunity Scoring** lane.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the dataset
from pathlib import Path

path = Path("data/raw/content_refresh_anonymized.csv")

if not path.exists():
    path = Path("content_refresh_anonymized.csv")

df = pd.read_csv(path)
# Basic statistics
total_pages = len(df)
down_pct = (df["trend_direction"] == "down").mean() * 100
older_than_year_pct = (df["content_age_days"] > 365).mean() * 100

print(f"Total pages: {total_pages:,}")
print(f"Pages with downward trend: {down_pct:.1f}%")
print(f"Pages older than one year: {older_than_year_pct:.1f}%")
```
